```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# 加载已有量子电路
# qasm_path = "/Users/AntiEntropy/Documents/Research/Low-energy Simulation/Numerical/f3c/examples"
# qasm_path += "/Ising_circuit/N5/TFXY_compressed_circuit_t_0.1.qasm"

qasm_path = "./qasm2/evolution_circuit_qasm2.qasm"
qc = QuantumCircuit.from_qasm_file(qasm_path)

# 1. 创建一个初始态，这里以 |0...0⟩ 初始态为例
initial_state = Statevector.from_label('0' * qc.num_qubits)

# 2. 将量子电路作用于初始态上
final_state = initial_state.evolve(qc)

# 3. 输出最终态的向量
print(final_state)
```
